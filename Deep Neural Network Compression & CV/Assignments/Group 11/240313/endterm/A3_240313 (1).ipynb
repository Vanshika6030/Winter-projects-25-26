````python
#Task I
import numpy as np

def convolve(img, kernel):

    # Flip kernel for true convolution
    kernel = np.flipud(np.fliplr(kernel))

    H, W = img.shape
    kH, kW = kernel.shape

    # Padding sizes
    pad_h = kH // 2
    pad_w = kW // 2

    # Reflect padding
    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    # Output image
    output = np.zeros((H, W), dtype=np.float64)

    # Convolution
    for i in range(H):
        for j in range(W):
            window = padded[i:i+kH, j:j+kW]
            output[i, j] = np.sum(window * kernel)

    return output
````

````python
#Task II
import matplotlib.pyplot as plt

def average_blur_gray(img, ksize):
    assert ksize % 2 == 1, "ksize must be odd"

    H, W = img.shape
    pad = ksize // 2

    kernel = np.ones((ksize, ksize), dtype=np.float64) / (ksize * ksize)
    padded = np.pad(img, ((pad, pad), (pad, pad)), mode="reflect")

    out = np.zeros((H, W), dtype=np.float64)

    for i in range(H):
        for j in range(W):
            window = padded[i:i+ksize, j:j+ksize]
            out[i, j] = np.sum(window * kernel)

    return out

img_path = "/content/drive/MyDrive/Screenshot 2025-12-15 131644.png"
img = plt.imread(img_path)

if img.ndim == 3:
    img = img.mean(axis=2)

img = img.astype(np.float64)

blurred = average_blur_gray(img, ksize=5)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(blurred, cmap="gray")
plt.title("Average Blurred Image")
plt.axis("off")

plt.show()

def average_blur_rgb(img, ksize):
    assert ksize % 2 == 1, "ksize must be odd"

    # Handle alpha channel if present
    if img.ndim == 3 and img.shape[2] == 4:
        img = img[:, :, :3]

    H, W, C = img.shape
    assert C == 3, "Input must be an RGB image"

    img = img.astype(np.float64)
    blurred = np.zeros_like(img, dtype=np.float64)

    for c in range(3):
        blurred[:, :, c] = average_blur_gray(img[:, :, c], ksize)

    return blurred

img_path = "/content/drive/MyDrive/Screenshot 2025-11-15 172059.png"

rgb_img = plt.imread(img_path)

# Remove alpha channel if exists
if rgb_img.ndim == 3 and rgb_img.shape[2] == 4:
    rgb_img = rgb_img[:, :, :3]
rgb_blur = average_blur_rgb(rgb_img, ksize=7)
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(np.clip(rgb_img, 0, 1))
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(np.clip(rgb_blur, 0, 1))
plt.title("Average Blurred RGB")
plt.axis("off")

plt.show()


````

````python
#Task III
def gaussian_kernel(ksize, sigma):
    assert ksize % 2 == 1, "ksize must be odd"

    ax = np.arange(-(ksize // 2), ksize // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)

    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel = kernel / np.sum(kernel)   # normalize

    return kernel

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)

def gaussian_blur_rgb(img, ksize, sigma):
    assert ksize % 2 == 1, "ksize must be odd"


    if img.ndim == 3 and img.shape[2] == 4:
        img = img[:, :, :3]

    img = img.astype(np.float64)

    H, W, C = img.shape
    assert C == 3, "Input must be RGB"

    out = np.zeros_like(img, dtype=np.float64)

    kernel = gaussian_kernel(ksize, sigma)

    # Apply convolution on each channel
    for c in range(3):
        out[:, :, c] = convolve(img[:, :, c], kernel)

    return out

img_path = "/content/drive/MyDrive/Screenshot 2025-11-15 172059.png"

img = plt.imread(img_path)

rgb_img = img.copy()

if img.ndim == 3:
    gray_img = img[:, :, :3].mean(axis=2)
else:
    gray_img = img

gray_img = gray_img.astype(np.float64)
ksize = 7
sigma = 1.5

gray_blur = gaussian_blur_gray(gray_img, ksize, sigma)
rgb_blur  = gaussian_blur_rgb(rgb_img, ksize, sigma)
plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray_img, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray_blur, cmap="gray")
plt.title("Gaussian Blurred Grayscale")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(np.clip(rgb_img, 0, 1))
plt.title("Original RGB")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(np.clip(rgb_blur, 0, 1))
plt.title("Gaussian Blurred RGB")
plt.axis("off")

plt.show()

````

````python
sobel_x = np.array([
    [-1,  0,  1],
    [-2,  0,  2],
    [-1,  0,  1]
], dtype=np.float64)

sobel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float64)
img_path = "/content/drive/MyDrive/Screenshot 2025-11-15 172059.png"

img = plt.imread(img_path)

if img.ndim == 3:
    img = img[:, :, :3].mean(axis=2)

img = img.astype(np.float64)
Gx = convolve(img, sobel_x)
Gy = convolve(img, sobel_y)
gradient_magnitude = np.sqrt(Gx**2 + Gy**2)
grad_norm = (gradient_magnitude / gradient_magnitude.max()) * 255
grad_norm = grad_norm.astype(np.uint8)

threshold = 50

edge_map = np.zeros_like(grad_norm)
edge_map[grad_norm > threshold] = 255
plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(Gx, cmap="gray")
plt.title("Sobel Gx")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(Gy, cmap="gray")
plt.title("Sobel Gy")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(grad_norm, cmap="gray")
plt.title("Gradient Magnitude (Normalized)")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(edge_map, cmap="gray")
plt.title("Edge Map (Thresholded)")
plt.axis("off")

plt.show()
````
